# EDSS 우선순위 미연결 ID 학교이력 검증

통폐합·폐교·명칭 변경·학위과정 중단을 구분해 우선순위 6개 ID의 검증 결과를 검토한다. 후보 ID는 자동 확정 매핑이 아니라 구조적 증거다.

```python
from pathlib import Path
import csv
import json

repo = Path.cwd().resolve()
if repo.name == 'notebooks':
    repo = repo.parent
validation_path = repo / 'data/metadata/edss_priority_school_history_validation.csv'
candidate_path = repo / 'data/metadata/edss_priority_school_history_candidates.json'
validation = list(csv.DictReader(validation_path.open(encoding='utf-8-sig')))
candidates = json.loads(candidate_path.read_text(encoding='utf-8'))
len(validation), len(candidates['targets'])
```

```python
[{k: row[k] for k in ('year', 'open_id', 'validation_status', 'likely_entity', 'institution_change_type', 'safe_join_action')} for row in validation]
```

```python
[(target['open_id'], target['departments'], target['exact_row_match_ids'][:10]) for target in candidates['targets']]
```
